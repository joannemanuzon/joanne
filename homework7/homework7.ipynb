```python
import seaborn as sns
```

```python
df2 = sns.load_dataset('planets')
print(df2)
```

```python
df2_clean = df2.dropna()
print(df2_clean)
```

```python
df2_clean.shape
# 498 rows, 6 columns
```

```python
df2_clean.head()
```

```python
df2_clean.tail()
```

```python
# Method: Method in which planet was found
# Number: Number of exoplanets in system found through method listed
# Orbital period: Orbital period of the exoplanet
# Mass: Mass of exoplanet
# Distance: Distance of exoplanet in light years
# Year: Year exoplanet was discovered
```

```python
# Subplots:
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Scatterplot:
sns.scatterplot(data=df2_clean, x="mass", y="distance", color="royalblue", ax=axes[0][0], label="Planets")
axes[0][0].set_title("Mass vs Distance")
axes[0][0].set_xlabel("Mass (Jupiter masses)")
axes[0][0].set_ylabel("Distance (light years)")
axes[0][0].legend()

sns.scatterplot(data=df2_clean, x="orbital_period", y="year", color="darkorange", ax=axes[0][1])
axes[0][1].set_title("Orbital Period vs Year")
axes[0][1].set_xlabel("Orbital Period (days)")
axes[0][1].set_ylabel("Year of Discovery")

# Histogram:
sns.histplot(data=df2_clean, x="mass", color="seagreen", ax=axes[1][0])
axes[1][0].set_title("Mass Distribution")
axes[1][0].set_xlabel("Mass (Jupiter masses)")
axes[1][0].set_ylabel("Count")

# Boxplot:
sns.boxplot(data=df2_clean, x="method", y="distance", palette="coolwarm", ax=axes[1][1])
axes[1][1].set_title("Distance by Detection Method")
axes[1][1].set_xlabel("Detection Method")
axes[1][1].set_ylabel("Distance (light years)")
axes[1][1].tick_params(axis='x', rotation=90)

plt.tight_layout()
plt.show()
```

```python
# Mass vs. Distance: Shows how each planet's mass relates to the distance of its star system from Earth. I chose these columns because they are core
#                    physical properties, and together they reveal whether heavier planets tend to be found farther away. From the plot, I learned
#                    that most detected planets are relatively close and there is no strong visible coorelation.
#
# Orbital Period vs. Year: Represents how a planet's orbital period varies with the year it was discovered. I chose these columns to explore how 
#                          improvements in detection technology affect the types of planets we find, and I learned that more recent discoveries 
#                          include many short-period planets
#
# Mass: Shows the distribution of planet masses in the dataset. I chose this column because a histogram clearly reveals how common different mass
#       ranges are, and I learned that most planets have masses below Jupiter's but a few extremely massive planets create a long tail. 
#
# Distance by Detection Method: Compares the distances of discovered planets across different detection methods. I chose these columns to see how
#                               each method performs in terms of how far away it can detect planets, and I learned that some methods—like radial
#                               velocity—tend to find planets at greater distances than others. 
```
