```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("exoplanet_data.csv" , comment = "#")

print(df.shape)
# 597 rows, 92 columns
```

```python
print(df.columns)

# pl_insollim, ttv_flag, rastr, ra, ttv_flag
```

```python
# Table Label: Planet Name
# Database Column: pl_name
# Description: Expolanet name
# This is the planet's official name used by astronomers

# Table Label: Host Name
# Database Column: hostname
# Description: Name of the star that the planet orbits 
# This is the star that the exoplanet revolves around

# Table Label: Discovery Method 
# Database Column: discoverymethod
# Description: Method by which the planet was discovered
# This tells us how the astronomer detected the planet, such as by the transit method or the radial velocity method

# Table Label: Spectral Type
# Database Column: st_spectype
# Description: Spectral type of the host star
# This describes the type of star based on its temperature and light spectrum

# Table Label: Equilibrium Temperature (K)
# Database Column: pl_eqt
# Description: Planet equilibrium temperature
# This is an estimate of how hot or cold the planet is based on the energy it receives from its star

# Table Label: Planet Mass
# Database Column: pl_bmassj
# Description:  Best available planet mass estimate in Jupiter masses
# This tells us how massive the planet is compared to Jupiter 
```

```python
df2 = df[[ "pl_name", 
                "hostname",
                "discoverymethod",
                "st_spectype",
                "pl_eqt",
                "pl_bmassj"
              ]]

print(df2)
```

```python
df2 = df2.dropna()

print(df2.shape)
# 75 rows, 6 columns
```

```python
df2.value_counts("discoverymethod")

# Transit, Radial Velocity, and Image
# 60 Transit, 13 Radial Velocity, 2 Imaging. Most is Transit, imaging has the least.

# Transit = Measuring the slight dimming of a star's light as a planet passes in front of it
# Radial Velocity = Measured using the Doppler Effect
# Imaging = Capturing photographs of the planetsT
```

```python
plt.figure(figsize=(8, 6))

for method in df2["discoverymethod"].unique():
    subset = df2[df2["discoverymethod"] == method]

    plt.scatter(
        subset["pl_bmassj"],    
        subset["pl_eqt"],         
        label=method
    )

plt.title("Planet Mass vs Equilibrium Temperature")
plt.xlabel("Planet Mass (Jupiter Masses)")
plt.ylabel("Equilibrium Temperature (K)")
plt.legend()

plt.show()
```

```python
plt.figure(figsize=(8, 6))

for method in df2["discoverymethod"].unique():
    subset = df2[df2["discoverymethod"] == method]

    plt.scatter(
        np.log(subset["pl_bmassj"]),    
        np.log(subset["pl_eqt"]),         
        label=method
    )

plt.title("Log Planet Mass vs Equilibrium Temperature")
plt.xlabel("log(Planet Mass (Jupiter Masses))")
plt.ylabel("log(Equilibrium Temperature (K))")
plt.legend()

plt.show()
```

```python
# Larger planets tend to be found using imaging
```

```python
coldest = df2[df2["pl_bmassj"] > 5]

print(coldest[["pl_name"]])

# Name: eps Ind A b
# Index 593
# Equilibrium Temperature: 2240.0K
# 15 Jupiter Masses
# Orbits USco1556 A

print(coldest[["st_spectype"]])
# M-type main sequence star, not similar to Sun
```

```python
coldest = df2[df2["pl_bmassj"] > 5]\
    .sort_values("pl_eqt") \
    .head(1)

plt.figure(figsize=(8, 6))

for method in df2["discoverymethod"].unique():
    subset = df2[df2["discoverymethod"] == method]

    plt.scatter(
        np.log(subset["pl_bmassj"]),    
        np.log(subset["pl_eqt"]),         
        label=method
    )

plt.scatter(
    np.log(coldest["pl_bmassj"]),
    np.log(coldest["pl_eqt"]),
    color="red",
    marker="*",
    s=300,
    label=coldest["pl_name"].iloc[0]
)

plt.title("Log Planet Mass vs Equilibrium Temperature")
plt.xlabel("log(Planet Mass (Jupiter Masses))")
plt.ylabel("log(Equilibrium Temperature (K))")
plt.legend()

plt.show()
```
