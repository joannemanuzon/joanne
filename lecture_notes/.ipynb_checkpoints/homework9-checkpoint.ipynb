```python
print(df.columns)
# x, y_noise_1, y_noise_2, y_noise_3, y_noise_4, y_noise_5
```

```python
plt.scatter(df['x'], df['y_noise_1'])
plt.xlabel("x")
plt.ylabel("y_noise_1")
plt.title("x vs. y_noise_1")
plt.legend()

plt.show()
```

```python
def model(x, A, k):
    return A * np.exp(-k * x)
```

```python
p0 = [1,1]
popt, pcovt = curve_fit(model, df['x'], df['y_noise_1'], p0 = p0)
A_fit, k_fit = popt
print(f"Fit: y = {A_fit:.2f} * e^(-{k_fit:.2f} * x)")
```

```python
residuals = df['y_noise_1'] - model(df['x'], A_fit, k_fit)
chi2 = np.sum(residuals ** 2)
dof = len(df['x']) - len(popt)
chi2_reduced = chi2 / dof
```

```python
import matplotlib.pyplot as plt
import pandas as pd

plt.scatter(df['x'], df['y_noise_1'])
plt.plot(df['x'], A_fit * df['x'] + k_fit, "r", label = "fitted model")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```
