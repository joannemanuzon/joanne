```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y)
plt.title("Scattered Points")
plt.show()
```

```python
x = np.linspace(0, 2*np.pi, 200)

fig, axes = plt.subplots(1, 3)

axes[0].plot(x, np.sin(x))
axes[0].set_title("Function of sin(x)")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")

axes[1].plot(x, np.sin(2 * x))
axes[1].set_title("Function of sin(2x)")
axes[1].set_xlabel("x")
axes[1].set_ylabel("y")

axes[2].plot(x, np.sin(3 * x))
axes[2].set_title("Function of sin(3x)")
axes[2].set_xlabel("x")
axes[2].set_ylabel("y")

```

```python
x = np.linspace(0, 2 * np.pi, 200)



a_vals = [0.2, 0.5, 1, 2, 10]



plt.figure()

for a in a_vals:
    plt.plot(x, x ** a, label = f"x^{a}")

plt.legend()
plt.title("Function of x^a")
plt.xlabel("x")
plt.ylabel("x^a")
plt.show

```

```python
def generateplots(name):
    current_func = getattr(np, name)

    x = np.linspace(-10, 10, 400)

    y = current_func(x)

    plt.figure()
    plt.plot(x, y)
    plt.title(f"Function of {name}(x)")
    plt.xlabel("x")
    plt.ylabel(f"{name}(x)")
    plt.show()

generateplots("sin")
generateplots("tan")
    
```
