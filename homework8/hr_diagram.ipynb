```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt(r"C:\Users\jmanu\pythondecalfa\course_assignments\homework7\star_data.txt", skiprows = 1)

print(data.shape)
# 2642 rows
```

```python
magnitude = data[:,0]
temperature = data[:,1]
```

```python
print("Minimum Magnitude:", magnitude.min())
print("Index of Star:", magnitude.argmin())

print("Maximum Magnitude:", magnitude.max())
print("Index of Star:", magnitude.argmax())
```

```python
print("Minimum Temperature", temperature.min())
print("Index of Star", temperature.argmin())

print("Maximum Temperature", temperature.max())
print("Index of Star", temperature.argmax())
```

```python
print("Brightest Star Index:", magnitude.argmin())
print("Dimmest Star Index:", magnitude.argmax())
```

```python
# 3.3
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize = (8, 6))

plt.scatter(log_temperature, log_magnitude, label = "Stars")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.title("Hertzsprung-Russel Diagram")

brightest = magnitude.argmin()
dimmest = magnitude.argmax()

hottest = temperature.argmax()
coolest = temperature.argmin()

plt.scatter(log_temperature[brightest],
            log_magnitude[brightest],
            color = "black", 
            marker = "*",
            s = 300,
            label = "Brightest Star")

plt.scatter(log_temperature[dimmest],
            log_magnitude[dimmest],
            color = "purple",
            marker = "*",
            s = 300,
            label = "Dimmest Star")

plt.scatter(log_temperature[hottest],
            log_magnitude[hottest],
            color = "green", 
            marker = "*",
            s = 300,
            label = "Hottest Star")

plt.scatter(log_temperature[coolest],
            log_magnitude[coolest],
            color = "orange",
            marker = "*",
            s = 300,
            label = "Coolest Star")

plt.legend()

plt.show()
```
