```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [7, 10, 13]
])

def is_prime(num):
    if num < 2:
        return False
    for i in range (2, num):
        if num % i == 0:
            return False
    return True

def rowswithprime(array):
    newrows = []
    for row in array:
        for num in row:
            if is_prime(num):
                newrows.append(row)
    return np.array(newrows)

print(rowswithprime(arr))
```

```python
def checkerboard():
    arr = np.zeros((8,8))
    
    for i in range(8):
        if i % 2 == 0:
            arr[i, ::2] = 1
        else:
            arr[i, 1::2] = 1
    return arr

print(checkerboard())
```

```python
import numpy as np

def reverse_checkerboard():
    board = np.zeros((8, 8), dtype=int)

    for i in range(8):
        if i % 2 == 0:
            board[i, 1::2] = 1 
        else:
            board[i, ::2] = 1 

    return board

print(reverse_checkerboard())
        
```

```python
def expansion(words, spaces):
    gap = ' ' * spaces
    expanded = []
    
    for word in words:
        new_word = ""
        for char in word:
            new_word += char + gap
        new_word = new_word.rstrip()
        expanded.append(new_word)
    
    return np.array(expanded)

universe = np.array(['galaxy', 'clusters'])
print(expansion(universe, 1))
print(expansion(universe, 2))
```

```python
import numpy as np

def secondDimmest(stars):
    # Sort each row in descending order (brightest first)
    sorted_stars = np.sort(stars, axis=1)[:, ::-1]
    
    # Pick the second value (index 1) from each row
    second_brightest = sorted_stars[:, 1]
    
    return second_brightest

# Example usage
np.random.seed(123)
stars = np.random.randint(500, 2000, (5, 5))
print(stars)
print(secondDimmest(stars))
```
